# Question 3.  Which areas have the fastest growing and shrinking crime rates.

Period available:  Nov 2015 to Nov 2017

```python
df['month'] = df['OCCURRED ON'].dt.to_period('M')
df['month'].unique()
```

```python
df['month'].value_counts().sort_index()
```

Crimes per zip per month:

```python
trend_zip = (
    df.groupby(['ZIP', 'month'])
      .size()
      .reset_index(name='count')
      .sort_values(['ZIP', 'month'])
)
trend_zip
```

Gemini offered to run code for me if I explained what I wanted.  This was my input:  "Identify the areas with the fastest growing and shrinking crime rates.  Calculate month to month percent changes in crime counts for each ZIP.  Average the change over 11/2015 to 11/2017. Sort by descending values."  ZIP codes with the highest average monthly percent increases represent the fastest-growing areas, while ZIPs with the most negative average changes represent the fastest-shrinking ZIPs.

```python
# Calculate month-to-month percentage change for each zip code
trend_zip['pct_change'] = trend_zip.groupby('ZIP')['count'].pct_change()

# Remove the first month for each zip code as it will have NaN for percent change
trend_zip_filtered = trend_zip.dropna(subset=['pct_change']).copy()

# Filter out zip codes with a low total number of crimes across the period
# (e.g., less than a certain threshold, let's use 50 as a starting point)
zip_counts = trend_zip_filtered.groupby('ZIP')['count'].sum()
zips_to_keep = zip_counts[zip_counts >= 50].index

trend_zip_filtered = trend_zip_filtered[trend_zip_filtered['ZIP'].isin(zips_to_keep)]

# Calculate the average monthly percentage change for each zip code
average_pct_change = trend_zip_filtered.groupby('ZIP')['pct_change'].mean().sort_values(ascending=False)

print("ZIP codes with the fastest growing crime rates (average monthly % change):")
display(average_pct_change.head())

print("\nZIP codes with the fastest shrinking crime rates (average monthly % change):")
display(average_pct_change.tail())
```

The analysis shows that crime is rising most rapidly in ZIP codes 85251, 85087, 85045, 85083, and 85331, with average monthly increases ranging from +18% to +34%. ZIP codes 85035, 85029, 85023, 85031, and 85042 exhibit small but steady monthly decreases in crime (around −1–2%) over the same time period.

# Question 4.  Group crimes by violent and non-violent:

Clean up first, and define:

```python
df['UCR CRIME CATEGORY'] = (
    df['UCR CRIME CATEGORY']
      .str.strip()
      .str.replace('"', '', regex=False)
)
```

```python
violent = [
    "AGGRAVATED ASSAULT",
    "ROBBERY",
    "RAPE",
    "MURDER AND NON-NEGLIGENT MANSLAUGHTER"
]

```

```python
df['VIOLENCE'] = df['UCR CRIME CATEGORY'].isin(violent).map({
    True: 'Violent',
    False: 'Non-Violent'
})
df['UCR CRIME CATEGORY'].unique()
```

```python
df['VIOLENCE'].value_counts()
```
